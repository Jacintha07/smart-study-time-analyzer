```python
logs=[]
n=int(input("How many study sessions? "))
for i in range(n):
    day=input("Day (Mon/Tue?...): ")
    subject=input("Subject:")
    hours=float(input("Hours studied: "))
    logs.append((day,subject,hours))
print(logs)
total_hours=0
for item in logs:
    total_hours=total_hours+item[2]
print("Total study Hours:",total_hours)
python_hours=0
maths_hours=0
english_hours=0
for item in logs:
    if item[1]=="Python":
        python_hours+=item[2]
    elif item[1]=="Maths":
        maths_hours+=item[2]
    elif item[1]=="English":
        english_hours+=item[2]
print("PYTHON HOURS:",python_hours)
print("MATHS HOURS:",maths_hours)
print("ENGLISH HOURS:",english_hours)
most_subject="Python"
most_hours=python_hours
least_subject="Python"
least_hours=python_hours
if maths_hours>python_hours:
    most_subject="Maths"
    most_hours=maths_hours
if english_hours>maths_hours:
    most_subject="English"
    most_hours=english_hours
if least_hours>maths_hours:
    least_subject="Maths"
    least_hours=maths_hours
if least_hours>english_hours:
    least_subject="English"
    least_hours=english_hours
average=total_hours/len(logs)
print("Most studied:",most_subject,"-",most_hours)
print("Least studied:",least_subject,"-",least_hours)
print("Average Per session:", average)
percent=(most_hours/total_hours)*100
print("Insight: You spend the most time on", most_subject, "-", round(percent), "% of your total study time")
```
